```python
import matplotlib.pyplot as plt
import pandas as pd
data = pd.read_csv("assignment3.csv")
data
```

```python
x=data.iloc[:,:-1]
y=data.iloc[:,-1]
y 
```

```python
x['Age']=x['Age'].map({'Youth':1,'Middle':2,'Senior':3})
x['Income']=x['Income'].map({'High':1, 'Medium':2, 'Low':3})
x['Student']=x['Student'].map({'No':0, 'Yes':1})
x['Creditrating']=x['Creditrating'].map({'Fair':1, 'Excellent':2})
x 
```

```python
from sklearn.naive_bayes import GaussianNB
nb=GaussianNB()
nb.fit(x,y)
print(nb.predict([[3,3,1,1]]))
```
