```python
GermanCars = pd.DataFrame({"Company":["Ford","Mercedes","BMW","Audi"],"Price":[23845,171995,135925,71400]})
JapaneseCars = pd.DataFrame({"Company":["Toyota","Honda","Nissan","Mitubishi"],"Price":[29995,23600,61500,58900]})

data = pd.concat([GermanCars,JapaneseCars],keys=['Germany','Japan'])
print(data)
```

```python
Car_Price = pd.DataFrame({"Company":["Toyota","Honda","BMW","Audi"],"Price":[23845,17995,135925,71400]})
car_Horsepower = pd.DataFrame({"Company":["Toyota","Honda","BMW","Audi"],"horsepower":[141,80,182,160]})

data = pd.merge(Car_Price,car_Horsepower,on="Company")

print(data)
```
